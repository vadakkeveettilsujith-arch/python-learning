# Python Data Types

Python has several built-in data types used to store different kinds of information.

Here we will discuss:

- String
- Float
- Integer
- Boolean
- Complex numbers

## FUNCTION USED TO DETERMINE THE DATA TYPE
- We use type() function to determine the datatype defined in a variable
-output are as follows:
   - `int` (integer)is used for whole numbers.
   - `float` (float)is used for decimal numbers.
   - `str`(string) is used for text.
   - `bool`(boolean) represents `True` or `False`
   - `complex`(complex) represents complex numbers

```python
#example of type() fuction 
name = 'sujith'
print(type(name))
age=25
print(type(age))
```

## STRINGS

- This are data types used for text.
- Kept inside quotations (double,single,triple single,triple double).
- example are shown below.


```python
#strings_examples
name="Sujith"
place='Wayanad'
qualification="""MSc Physics"""
current_location='''Eranakulam'''
print(name,place,qualification,current_location)
```

## INTEGER

- This data types are used for whole numbers.
- No quotations used

```python
#Integer_example
age=25
register_num=123456
print('age : ',age)
print('Reg.No. : ',register_num)
```

## FLOAT
- This data type includes numberes with decimal point.
- Here also we won't use quotations

```python
#float_example
height=169.55
weight=80.10
print('height(in cm): ',height)
print('weight(in kg) : ',weight)
print(type(height))
```

## BOOLEAN
- Boolean is a data type that has two possible values: True or False.
- Boolean values are commonly used in conditional statements and logical operations.
- In Python, True and False are Boolean values and must begin with a capital letter.

```python
#example of boolean data type
is_student = True
is_employed = False

print(is_student)
print(type(is_student))
#conditional_type
age = 25
verification=age>=18
print(verification)
```

## COMPLEX NUMBERS
- In this data type we represent complex numbers which have a real and imaginary part
- eg: 3+2j where 3 is the real part and 2j is the imaginary part

```python
#example of complex data_type
cmp_num=3+5j
print(type(cmp_num))
```
