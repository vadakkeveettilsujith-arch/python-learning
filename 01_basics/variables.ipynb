## Python_variables

variables:Temporary storage location of data values in Python.

A variable is created when a value is assigned to it.

```python
#creating variables and printing
name='sujith'
age=25
is_student=True

print(name)
print(age)
print(is_student)
```

## Multiple Variable Assignment

Python allows multiple variables to be assigned in a single statement.


```python
name, age, city = "Sujith", 25, "Kerala"

print("Name:", name)
print("Age:", age)
print("City:", city)
```

## Changing the Value of a Variable

A variable can be reassigned to a different value.

```python
age = 24
print("Age:", age)

age = 25
print("Updated age:", age)
```
